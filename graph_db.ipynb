# Graph RAG with Neo4j Aura

This notebook builds a small GraphRAG pipeline with LangChain, Groq, and Neo4j Aura.

## Workflow

1. Load connection settings from `.env`.
2. Prepare and chunk source text.
3. Extract LangChain `GraphDocument` objects.
4. Persist the existing graph documents into Neo4j with idempotent transactions.

Credentials are loaded from environment variables and are never hard-coded in the notebook.

## Graph extraction

`LLMGraphTransformer` converts each document chunk into a `GraphDocument` containing typed nodes, relationships, and the source `Document`. The persistence stage below uses this existing in-memory collection and does not invoke the LLM again.

## Neo4j persistence

The next cells use the existing `graph_documents` collection. They do not call the LLM or regenerate the graph.

### Data model

- `GraphEntity` is the shared base label for extracted nodes.
- The LangChain node `id` is preserved as the Neo4j `id` property.
- The LangChain node `type` becomes both a label and the `entity_type` property.
- Relationships are merged by source ID, target ID, and relationship type.
- `SourceDocument` nodes retain source metadata as JSON.
- `EXTRACTED_FROM` links entities back to their source documents.

### Operational guarantees

- Values use parameterized Cypher.
- Each `GraphDocument` is written in a transaction.
- `MERGE` makes rerunning the persistence cell idempotent where IDs and types match.
- The driver is closed in a `finally` block, including after an insertion error.

```python
import hashlib
import json
import re
from typing import Any

from neo4j import GraphDatabase


def _neo4j_value(value: Any) -> Any:
    """Convert LangChain property values to Neo4j-compatible property values."""
    if value is None or isinstance(value, (str, int, float, bool)):
        return value
    if isinstance(value, dict):
        return json.dumps(value, ensure_ascii=True, default=str)
    if isinstance(value, (list, tuple)):
        converted = [_neo4j_value(item) for item in value]
        if any(isinstance(item, str) and item.startswith("{") for item in converted):
            return json.dumps(value, ensure_ascii=True, default=str)
        return converted
    return str(value)


def _properties(properties: dict[str, Any] | None) -> dict[str, Any]:
    return {key: _neo4j_value(value) for key, value in (properties or {}).items()}


def _identifier(value: str, fallback: str) -> str:
    """Allow only simple Cypher identifiers before inserting labels/types in a query."""
    candidate = re.sub(r"[^A-Za-z0-9_]", "_", str(value)).strip("_")
    if not candidate or not re.match(r"^[A-Za-z_]", candidate):
        return fallback
    return candidate


def _source_properties(source: Any) -> tuple[str, dict[str, Any]]:
    metadata = getattr(source, "metadata", {}) or {}
    content = getattr(source, "page_content", "") or ""
    source_key = hashlib.sha256(
        json.dumps(
            {"metadata": metadata, "content": content},
            sort_keys=True,
            default=str,
        ).encode("utf-8")
    ).hexdigest()
    return source_key, {
        "source_id": source_key,
        "metadata": json.dumps(metadata, ensure_ascii=True, default=str),
    }
```

```python
def write_graph_document(tx, graph_document) -> None:
    source_id, source_properties = _source_properties(graph_document.source)

    tx.run(
        """
        MERGE (source:SourceDocument {source_id: $source_id})
        SET source += $source_properties
        """,
        source_id=source_id,
        source_properties=source_properties,
    ).consume()

    for node in graph_document.nodes:
        label = _identifier(node.type, "GraphNode")
        node_properties = _properties(node.properties)
        node_properties["id"] = node.id
        node_properties["entity_type"] = node.type

        query = f"""
        MERGE (entity:GraphEntity {{id: $id}})
        SET entity:`{label}`
        SET entity += $properties
        MERGE (entity)-[:EXTRACTED_FROM]->(source:SourceDocument {{source_id: $source_id}})
        """
        tx.run(
            query,
            id=node.id,
            properties=node_properties,
            source_id=source_id,
        ).consume()

    for relationship in graph_document.relationships:
        relationship_type = _identifier(relationship.type, "RELATED_TO")
        query = f"""
        MATCH (source:GraphEntity {{id: $source_id}})
        MATCH (target:GraphEntity {{id: $target_id}})
        MERGE (source)-[relation:`{relationship_type}`]->(target)
        SET relation += $properties
        """
        tx.run(
            query,
            source_id=relationship.source.id,
            target_id=relationship.target.id,
            properties=_properties(relationship.properties),
        ).consume()
```

```python
graph_document_count = len(graph_documents)
node_count = sum(len(item.nodes) for item in graph_documents)
relationship_count = sum(len(item.relationships) for item in graph_documents)

print(f"GraphDocuments: {graph_document_count}")
print(f"Nodes to merge: {node_count}")
print(f"Relationships to merge: {relationship_count}")

neo4j_driver = GraphDatabase.driver(
    NEO4J_URI,
    auth=(NEO4J_USERNAME, NEO4J_PASSWORD),
)

try:
    neo4j_driver.verify_connectivity()
    print("Neo4j Aura connection verified")

    with neo4j_driver.session(database=NEO4J_DATABASE) as session:
        for index, graph_document in enumerate(graph_documents, start=1):
            session.execute_write(write_graph_document, graph_document)
            print(
                f"Inserted GraphDocument {index}/{graph_document_count} "
                f"({len(graph_document.nodes)} nodes, "
                f"{len(graph_document.relationships)} relationships)"
            )

        persisted_counts = session.run(
            """
            MATCH (entity:GraphEntity)
            WITH count(entity) AS nodes
            OPTIONAL MATCH ()-[relation]->()
            WHERE type(relation) <> 'EXTRACTED_FROM'
            WITH nodes, count(relation) AS graph_relationships
            OPTIONAL MATCH ()-[source_link:EXTRACTED_FROM]->()
            RETURN nodes, graph_relationships, count(source_link) AS source_links
            """
        ).single()

    print(
        "Successful insertion: "
        f"{persisted_counts['nodes']} unique GraphEntity nodes, "
        f"{persisted_counts['graph_relationships']} graph relationships, "
        f"{persisted_counts['source_links']} source links"
    )
finally:
    neo4j_driver.close()
    print("Neo4j driver closed")
```
